````python
import nfl_data_py as nfl
import pandas as pd
from datetime import datetime

# Configure seasons to download
SEASONS = [2024, 2023, 2022]  # Adjust as needed

print(f"Downloading data for seasons: {SEASONS}")
print(f"Started: {datetime.now()}\n")
````

## Download Schedules

````python
print("Downloading schedules...")
schedules = nfl.import_schedules(SEASONS)
schedules.to_csv('schedules.csv', index=False)
print(f"✓ Saved schedules.csv ({schedules.shape[0]} rows)")
````

## Download Snap Counts

````python
print("Downloading snap counts...")
snaps = nfl.import_snap_counts(SEASONS)
snaps.to_csv('snap_counts.csv', index=False)
print(f"✓ Saved snap_counts.csv ({snaps.shape[0]} rows)")
````

## Download Weekly Rosters

````python
print("Downloading weekly rosters...")
rosters = nfl.import_weekly_rosters(SEASONS)
rosters.to_csv('rosters_weekly.csv', index=False)
print(f"✓ Saved rosters_weekly.csv ({rosters.shape[0]} rows)")
````

## Download Files to Your Computer

````python
from google.colab import files
import os

# List all CSV and Parquet files
download_files = [f for f in os.listdir('.') if f.endswith(('.csv', '.parquet'))]

print(f"\nDownloading {len(download_files)} files to your computer...\n")

for filename in download_files:
    print(f"Downloading: {filename}")
    files.download(filename)

print("\n✓ All files downloaded!")
print(f"\nCompleted: {datetime.now()}")
````
